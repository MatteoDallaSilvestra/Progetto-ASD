```python
import pandas as pd
import matplotlib.pyplot as plt


try:
    df_raw = pd.read_csv("results.csv")
    print(f"Successfully loaded {len(df_raw)} total measurements.")
    
    display(df_raw.head())
    
except FileNotFoundError:
    print("Error: results.csv not found. Make sure to run main.py first!")
```

```python
# Impostiamo semplicemente la colonna 'n' come indice del DataFrame
# in modo che diventi automaticamente l'asse X del grafico
df_plot = df_raw.set_index('n')

display(df_plot.head())
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

df_plot.plot(kind='line', marker='.', ax=ax, linewidth=2, markersize=8)

ax.set_title('Medium Insert Time by Tree Type (Mediana)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Number of Nodes (n) - Logarithmic Scale', fontsize=12)
ax.set_ylabel('Median Insert Time (Seconds)', fontsize=12)

ax.set_xscale('log')

ax.grid(True, which="both", linestyle="--", linewidth=0.5, alpha=0.7)

ax.legend(title='Tree Type', fontsize=11, title_fontsize=12)

plt.savefig("grafico_comparativo.png", dpi=300, bbox_inches='tight')
print("Chart successfully saved as 'grafico_comparativo.png'")

plt.show()
```
